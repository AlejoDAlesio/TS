El objetivo del trabajo es analizar las ecuaciones en diferencias de determinados sistemas que representan un filtro de media móvil, vamos a calcular su $T(Z)$, su respuesta en frecuencia de módulo y fase, respaldando el resultado obtenido analiticamente a través de la función Numpy de Python.  

$$a) \quad y(n) = x(n-3) + x(n-2) + x(n-1) + x(n)$$

<img src="TS6-a.jpg" width="500">

Verifico el resultado obtenido de manera analítica utilizando código.

```python

import numpy as np
import matplotlib.pyplot as plt



num = [1, 1, 1, 1]
den = [1, 0, 0, 0]   

# Grilla de frecuencias en el círculo unitario
N = 1024
omega = np.linspace(-np.pi, np.pi, N)
z = np.exp(1j * omega)

# Evaluar T(z) = P(z) / Q(z)
H = np.polyval(num, z) / np.polyval(den, z)

modulo = np.abs(H)
fase   = np.angle(H)   # devuelve en radianes, rango (-π, π]

plt.figure(figsize=(8, 5))

plt.subplot(2, 1, 1)
plt.plot(omega/np.pi, modulo)
plt.ylabel("|T|")
plt.xticks([-1, -0.5, 0, 0.5, 1], ["-π", "-π/2", "0", "π/2", "π"])
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(omega/np.pi, fase)
plt.ylabel("∠T [rad]")
plt.xlabel("ω")
plt.xticks([-1, -0.5, 0, 0.5, 1], ["-π", "-π/2", "0", "π/2", "π"])
plt.grid(True)

plt.tight_layout()
plt.show()
```
